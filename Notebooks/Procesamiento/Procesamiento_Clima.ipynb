-  Descripción: En este notebook se realiza la imputación de datos faltantes en algunas columnas, con el obetivo de contruir una vista minable robusta para alimentar los modelos de machine Learning. Ademas se agrega información del clima para cada uno de los lotes especificados.
-  Salida: Vista Minable depurada , con información del clima lista para ser utilizada en los entreanmientos de algortimos de Machine Learning.


```python
# Librerias y dependencias
# ==========================================================================
import numpy as np
import pandas as pd
from statistics import median,mode
```

```python
# Lectura del datframe (Contiene info de controles, fertilizaciín y suelo)
# =================================================================
matrizFinal = pd.read_csv("../../Data/Gold/datasetFinal_soil.csv")
matrizFinal.head(5)
```

```python
#  Descripción de las columnas del dataset
# =================================================
matrizFinal.info()
```

```python
# conteo de la cantidad de nulos para c/u de las columnas
# ================================================================
matrizFinal.isnull().sum().sort_values(ascending=False).head(20)
```

*Nota*: En este caso como el porcentaje de valores nulos en la mayoria de columnas  es menor al 1% se decidio utilizar tecnicas estadisticas como: (moda , mediana, variznza ) para realizar dicha emputación de los datos faltantes.


```python
# Para material Genetico
matrizFinal.MATERIAL_GENETICO.value_counts()
```

```python
# Se calcula (moda, media )  para c/d variable segun corresponda
# ========================================================

# Material Genetico
mode_MG = mode(matrizFinal.MATERIAL_GENETICO)
print("Moda Material Gnetico: ", mode_MG)

# Tipo Siembra
mode_TS = mode(matrizFinal.TIPO_SIEMBRA)
print("La moda para la variable TIPO_SIEMBRA: ",mode_TS)

# Dias Emerger
median_DEE = median(matrizFinal.DIAS_EN_EMERGER)
print("La media parala variable DIAS_EN_EMERGER es: ",median_DEE)

# Dias Emerger a Florecer
median_DEEF = median(matrizFinal.DIAS_EN_EMERGER_A_FLORECER)
print("La media parala variable DIAS_EN_EMERGER es: ",median_DEEF)

# Dias Florecer a Cosechar
median_DEFC = median(matrizFinal.DIAS_EN_FLORECER_A_COSECHAR)
print("La media parala variable DIAS_EN_FLORECER_A_COSECHAR es: ",median_DEFC)

# Cultivo Anterior
mode_CA = mode(matrizFinal.CULT_ANT)
print("La moda parala variable CULT_ANT es: ",mode_CA)

# Profundiad moteados rasta
median_PMR =median(matrizFinal.PROFUND_MOTEADOS_RASTA)
print("La media para al variable PROFUND_MOTEADOS_RASTA es: ", median_PMR)
```

```python
# Rempleazo de c/u de los valores faltantes en cada columna
# ===================================================================================================
matrizFinal.MATERIAL_GENETICO[matrizFinal.MATERIAL_GENETICO.isnull()] = mode_MG
matrizFinal.TIPO_SIEMBRA[matrizFinal.TIPO_SIEMBRA.isnull()] = mode_TS
matrizFinal.DIAS_EN_EMERGER[matrizFinal.DIAS_EN_EMERGER.isnull()] = median_DEE
matrizFinal.DIAS_EN_EMERGER_A_FLORECER[matrizFinal.DIAS_EN_EMERGER_A_FLORECER.isnull()] = median_DEEF
matrizFinal.DIAS_EN_FLORECER_A_COSECHAR[matrizFinal.DIAS_EN_FLORECER_A_COSECHAR.isnull()] = median_DEFC
matrizFinal.CULT_ANT[matrizFinal.CULT_ANT.isnull()] = mode_CA
matrizFinal.PROFUND_MOTEADOS_RASTA[matrizFinal.PROFUND_MOTEADOS_RASTA.isnull()]= median_PMR
```

```python
# Verificación de la imputación de datos
# ================================================================
matrizFinal.isnull().sum().sort_values(ascending=True).head(20)
```

```python
# Se realiza un mergue entre la matriz Final y datos del clima
# ============================================================
dataframe_final = pd.merge(matrizFinal,df_clima,how = 'inner', on=["ID_EVENTO"])
```

```python
# Eliminamos Columnas por su baja varianza 0; Todos los valores son iguales
# - Causan ruido innecesario en el modelo
# - No aportan información util
# - Se incluye tambien el id_evento que es un identificador que no aporta información
# df1_unmatched = matrizFinal[~matrizFinal['ID_EVENTO'].isin(dataframe_final['ID_EVENTO'])] 
borrarPorVarianza = ["ID_EVENTO", "ContEnfQui_Antes_Siem", "ContEnfQui_Siem_Eme","ContMalMec_Antes_Siem"
                    ,"FerOrg_Antes_Siem","FerOrg_Siem_Emer","Porc_ArF","Temp_Min_15_Freq_Veg",
                    "Temp_Min_15_Freq_For","Temp_Min_15_Freq_Mad","AREA"]
```

```python
# Verificacion de la varianza
df_varianza = dataframe_final[borrarPorVarianza].var()
df_varianza 
```

```python
# Se eliminan de la matriz Final las caracteristicas con alta varianza
dataframe_final = dataframe_final.drop(borrarPorVarianza,axis=1)
```
